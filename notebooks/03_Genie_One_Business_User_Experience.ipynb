### 🎯 What is Genie One?

Genie One is the **AI coworker for business users**. It provides:
- **Unified chat** — ask data questions in natural language
- **Asset discovery** — find dashboards, agents, and apps
- **Scheduled tasks** — automate recurring reports
- **Document drafting** — create shareable documents from chat
- **External sources** — connect Google Drive, Slack, SharePoint, etc.
- **Mobile access** — same experience on iOS and Android

**Key differentiator**: Genie One routes questions to the most relevant **Genie Agent**, dashboard, or metric view — users don't need to know which data source to query.

## ⏰ Cell 4: Scheduled Tasks

Scheduled tasks automate recurring analyses. They run at a specified time and post results to a chat thread — like having an AI analyst that works overnight.

### Example Scheduled Tasks for Manufacturing

| Task | Schedule | What It Does |
|---|---|---|
| Morning KPI Brief | Daily 7:00 AM | "Summarize yesterday's production KPIs across all plants. Flag any plant with yield <93% or uptime <85%." |
| Weekly Quality Report | Monday 8:00 AM | "Compare this week's quality pass rate to last week by plant. Highlight any plant with >2% decline." |
| Equipment Alert Digest | Daily 6:00 AM | "List all production lines with critical equipment alerts in the last 24 hours." |

### How to Create (UI Steps)

1. In Genie One chat, click **Scheduled Tasks** in the left sidebar
2. Click **+ New Task**
3. Enter the prompt (e.g., the Morning KPI Brief above)
4. Set the schedule (daily, weekly, etc.)
5. Click **Create**

The task runs automatically and posts results to a dedicated chat thread. On mobile, you'll receive a **push notification** (iOS).


```python
# Scheduled task templates for the PoC
scheduled_tasks = [
    {
        "name": "Morning KPI Brief",
        "prompt": "Summarize yesterday's production KPIs across all plants. Include total production volume, overall yield rate, quality pass rate, and equipment uptime. Flag any plant with yield below 93% or uptime below 85%.",
        "schedule": "Daily at 7:00 AM",
        "persona": "Plant Manager / Executive"
    },
    {
        "name": "Weekly Quality Report",
        "prompt": "Compare this week's quality pass rate to last week for each plant. Show the week-over-week change. Highlight any plant with more than 2% decline in pass rate. Include critical and major defect counts.",
        "schedule": "Every Monday at 8:00 AM",
        "persona": "Quality Director"
    },
    {
        "name": "Equipment Alert Digest",
        "prompt": "List all production lines that had critical equipment alerts in the last 24 hours. For each, show the plant name, line ID, sensor type, and number of critical readings. Sort by most alerts first.",
        "schedule": "Daily at 6:00 AM",
        "persona": "Maintenance Manager"
    },
]

print("⏰ SCHEDULED TASK TEMPLATES")
print("=" * 60)
print("Create these in Genie One → Scheduled Tasks → + New Task")
print()
for task in scheduled_tasks:
    print(f"📌 {task['name']}")
    print(f"   Schedule: {task['schedule']}")
    print(f"   Persona: {task['persona']}")
    print(f"   Prompt: {task['prompt'][:100]}...")
    print()
```

## 🔌 Cell 5: External Source Connections (MCP)

Genie One can connect to external tools via **MCP (Model Context Protocol)** connectors:

| Connector | What It Enables | Status |
|---|---|---|
| Google Drive | Search and reference documents, spreadsheets | GA |
| Gmail | Search emails for context | GA |
| Google Calendar | Check meeting schedules | GA |
| Slack | Search messages and channels | GA |
| Microsoft 365 | Teams, Outlook, SharePoint, OneDrive | GA |
| Atlassian | Jira issues, Confluence pages | GA |
| GitHub | Code repositories, issues, PRs | GA |

### Demo Scenario
"What does our Q3 operations strategy document say about the Asia Pacific expansion plan?"
→ Genie One searches Google Drive for the strategy doc AND queries the manufacturing data to show current APAC performance.

### How to Configure
1. In Genie One chat, click the **+** icon next to the message input
2. Select **Connect a source**
3. Choose the connector (e.g., Google Drive)
4. Authorize access
5. The connection persists across sessions
